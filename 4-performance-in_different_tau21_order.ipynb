```python
# Backend
import os

os.environ["KERAS_BACKEND"] = "jax"
```

```python
# Setup
import json
from pathlib import Path
from typing import cast

import numpy as np

from src.models import LearnableCutFlowModel
from src.utils import load_model, plot_learned_importance
```

```python
# Parameters: entered
n_runs = 10

# Path
FIGURES_DIR = Path("figures")
CHECKPOINTS_DIR = Path("checkpoints")
RESULTS_DIR = Path("results")
```

```python
# Parameters: derived
# Config
with open("1-dataset:config.json", "r") as f:
    dataset_config = json.load(f)

with open("2-model:config.json", "r") as f:
    model_config = json.load(f)

# Dataset
dataset_name = "real1"
features = dataset_config[dataset_name]["features"]
n_samples = dataset_config[dataset_name]["n_samples"]
seed = dataset_config[dataset_name]["seed"]

# Model
centers = [80, 0.15, 0.025, 2, 2, 0.3]
n_epochs = 200
batch_size = 512

# Figure
bins = [np.linspace(**kwargs) for kwargs in dataset_config[dataset_name]["bins"]]
```

```python
# feature: [0, 1, 2, 3, 4, 5]
feature_selection = [0, 1, 2, 3, 4, 5]

selection = "".join([str(i) for i in feature_selection])
prefix = f"4-dataset:real1-model:bdt,mlp,lcf-selection:{selection}-"
suffix = f"(x{n_runs})"
results_path = RESULTS_DIR / f"{prefix}results{suffix}.json"

with open(results_path, "r") as f:
    results = json.load(f)

# lcf_par
checkpoints = [
    load_model(file_path) for file_path in results["lcf_par"]["file_path"]["records"]
]
checkpoints = [cast(LearnableCutFlowModel, ckpt) for ckpt in checkpoints]

scores = []
for checkpoint in checkpoints:
    scores.append(checkpoint.importance.scores)

scores = np.array(scores)
scores_mean = np.mean(scores, axis=0)
scores_std = np.std(scores, axis=0)
baseline = checkpoints[0].importance.baseline
reordered_features = [features[i] for i in feature_selection]

plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
)
plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
    to_file=FIGURES_DIR / f"{prefix}-lcf_par-learned_importance{suffix}.pdf",
)
plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
    to_file=FIGURES_DIR / f"{prefix}-lcf_par-learned_importance{suffix}.png",
)

# lcf_seq
checkpoints = [
    load_model(file_path) for file_path in results["lcf_seq"]["file_path"]["records"]
]
checkpoints = [cast(LearnableCutFlowModel, ckpt) for ckpt in checkpoints]

scores = []
for checkpoint in checkpoints:
    scores.append(checkpoint.importance.scores)

scores = np.array(scores)
scores_mean = np.mean(scores, axis=0)
scores_std = np.std(scores, axis=0)
baseline = checkpoints[0].importance.baseline
reordered_features = [features[i] for i in feature_selection]

plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
)
plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
    to_file=FIGURES_DIR / f"{prefix}-lcf_seq-learned_importance{suffix}.pdf",
)
plot_learned_importance(
    scores=scores_mean,
    errors=scores_std,
    baseline=baseline,
    features=reordered_features,
    to_file=FIGURES_DIR / f"{prefix}-lcf_seq-learned_importance{suffix}.png",
)
```
