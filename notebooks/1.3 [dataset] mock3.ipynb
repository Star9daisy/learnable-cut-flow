```python
# Setup
from importlib import import_module

import numpy as np

from src.utils import (
    show_record_dataset,
    show_record_dataset_correlation,
    show_record_feature,
)
```

```python
# Parameters
dataset_name = "mock3"
n_samples = 200000
seed = 42

# For figures
bins = np.linspace(-10, 10, 101)
feature_names = ["x1", "x7", "x8"]
feature_names_latex = [r"$x_1$", r"$x_7$", r"$x_8$"]
```

```python
# Dataset
load_data = getattr(import_module(f"src.datasets.{dataset_name}"), "load_data")
(x_train, y_train), _ = load_data(n_samples, seed)
```

```python
# Figure: correlation
show_record_dataset_correlation(x_train, feature_names_latex)

# Save in png and pdf
show_record_dataset_correlation(
    x_train,
    feature_names_latex,
    to_file=f"figures/{dataset_name}_correlation.png",
)
show_record_dataset_correlation(
    x_train,
    feature_names_latex,
    to_file=f"figures/{dataset_name}_correlation.pdf",
)
```

```python
# Figure: distributions
show_record_dataset(x_train, y_train, bins, feature_names)

# Save in png and pdf
for i in range(x_train.shape[1]):
    show_record_feature(
        x_train,
        y_train,
        bins,
        feature_index=i,
        feature_name=feature_names_latex[i],
        to_file=f"figures/{dataset_name}_{feature_names[i]}.png",
    )
    show_record_feature(
        x_train,
        y_train,
        bins,
        feature_index=i,
        feature_name=feature_names_latex[i],
        to_file=f"figures/{dataset_name}_{feature_names[i]}.pdf",
    )
```
